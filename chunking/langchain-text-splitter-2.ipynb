```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
document = """
LangChain adalah framework open-source yang digunakan untuk membangun aplikasi berbasis Large Language Model (LLM).
Framework ini menyediakan abstraksi untuk prompt, memory, chains, agents, dan tools.

Dengan LangChain, developer dapat menghubungkan LLM dengan berbagai sumber data eksternal seperti database,
API, dokumen PDF, dan vector store seperti FAISS, Chroma, atau Elasticsearch.

Salah satu komponen penting di LangChain adalah text splitter.
Text splitter berfungsi untuk memecah dokumen panjang menjadi potongan-potongan kecil (chunks)
agar lebih mudah diproses oleh model bahasa.

RecursiveCharacterTextSplitter bekerja dengan memecah teks secara rekursif
berdasarkan separator seperti paragraf, baris baru, spasi, dan karakter.
Pendekatan ini membantu menjaga konteks semantik tetap utuh.

Text yang sudah di-split biasanya akan di-embed menggunakan embedding model
dan disimpan ke dalam vector database untuk keperluan retrieval (RAG).
"""
```

```python
text_splitter = RecursiveCharacterTextSplitter(chunk_size=100, chunk_overlap=0)
texts = text_splitter.split_text(document)
```

```python
from langchain_text_splitters import SentenceTransformersTokenTextSplitter
```

```python
splitter = SentenceTransformersTokenTextSplitter(chunk_overlap=0)
```

```python
text = "Lorem "
```

```python
count_start_and_stop_tokens = 2
```

```python
text_token_count = splitter.count_tokens(text=text) - count_start_and_stop_tokens
```

```python
print(text_token_count)
```

```python
token_multiplier = splitter.maximum_tokens_per_chunk // text_token_count + 1
```

```python
text_to_split = text * token_multiplier
```

```python
print(f"tokens in text to split: {splitter.count_tokens(text=text_to_split)}")
```

```python
text_chunks = splitter.split_text(text=text_to_split)

print(text_chunks[1])
```
